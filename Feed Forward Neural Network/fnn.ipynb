# **Feed Forward Neural Network**  
**A Feedforward Neural Network (FNN) is one of the most fundamental types of artificial neural networks—think of it as the "vanilla" version of deep learning**  
**Going to build simple Feed Forward Neural Network using tensorflow**

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist # pyright: ignore[reportMissingImports]
from tensorflow.keras.utils import to_categorical # pyright: ignore[reportMissingImports]
```

```python
tf.compat.v1.disable_eager_execution
```

**Load and preprocess MNIST data**

```python
(train_X, train_Y), (test_X, test_Y) = mnist.load_data()
train_X = train_X.reshape(-1, 784).astype('float32') / 255.0
test_X = test_X.reshape(-1, 784).astype('float32') / 255.0
train_Y = to_categorical(train_Y, 10)
test_Y = to_categorical(test_Y, 10)
```

**Network parameters**

```python
input_size = 784
hidden_size = 128
output_size = 10
num_epochs = 10
```

**Placeholders**

```python
X = tf.compat.v1.placeholder(tf.float32, shape=[None, input_size])
Y = tf.compat.v1.placeholder(tf.float32, shape=[None, output_size])
```

**Activation function**

```python
def swish(x):
    return x * tf.nn.sigmoid(x)
```

**Define layers using tf.keras.layers**

```python
hidden_layer = tf.keras.layers.Dense(units=hidden_size, activation=swish)(X)
dropout_layer = tf.keras.layers.Dropout(rate=0.3)(hidden_layer)
output_layer = tf.keras.layers.Dense(units=output_size)(dropout_layer)
```

**Loss and optimizer**

```python
loss = tf.reduce_mean(tf.nn.softmax_cross_entropy_with_logits(labels=Y, logits=output_layer))
optimizer = tf.compat.v1.train.AdamOptimizer(learning_rate=0.001)
train_op = optimizer.minimize(loss)
```

**Accuracy calculation**

```python
correct_pred = tf.equal(tf.argmax(output_layer, 1), tf.argmax(Y, 1))
accuracy_op = tf.reduce_mean(tf.cast(correct_pred, tf.float32))
```

**Initialize variables**

```python
init = tf.compat.v1.global_variables_initializer()
```

**Training variables**

```python
input_data = train_X
target_labels = train_Y
test_data = test_X
test_labels = test_Y
```

**Training loop and accuracy**

```python

with tf.compat.v1.Session() as sess:
    sess.run(init)
    for epoch in range(num_epochs):
        _, curr_loss = sess.run([train_op, loss], feed_dict={X: input_data, Y: target_labels})
        print(f"Epoch {epoch + 1}, Loss: {curr_loss:.4f}")
    accuracy = sess.run(accuracy_op, feed_dict={X: test_data, Y: test_labels})
    print(f"Test Accuracy: {accuracy:.4f}")
```
